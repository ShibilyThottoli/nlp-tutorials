```python
import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp("Dr. Strange Loves pav Bhaji of Mumbai . Hulk loves chat of Delhi")

for sentence in doc.sents:
    print(sentence)
```

```python
for sentence in doc.sents:
    for word in sentence:
        print(word)
```
